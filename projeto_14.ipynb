# Importando bibliotecas

# Limpeza e Transformação

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df['Quantidade'] = df['Quantidade'].replace('-', 0)
```

```python
df['Valor'] = df['Valor'].replace('-', 0)
```

```python
df['Quantidade'] = df['Quantidade'].astype(int)
```

```python
df['Valor'] = df['Valor'].astype(float)
```

```python
df.head()
```

```python
df.tail()
```

# Análise Exploratória

```python
df.describe()
```

```python
sns.boxplot(df['Quantidade'])
plt.title('Gráfico de caixa - Quantidade')
plt.ylabel('Quantidade')
plt.show()
```

```python
sns.boxplot(df['Valor'])
plt.title('Gráfico de caixa - Valor')
plt.ylabel('Valor')
plt.show()
```

```python
quantidade_media = df['Quantidade'].mean()
valor_medio = df['Valor'].mean()
```

```python
plt.figure(figsize=(6, 4))
plt.bar(['Quantidade Média'], [quantidade_media])
plt.title('Quantidade Média')
plt.ylabel('Média')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.bar(['Valor Médio'], [valor_medio])
plt.title('Valor Médio')
plt.ylabel('Média')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Quantidade'], bins=10, kde=True)
plt.title('Histograma - Quantidade')
plt.xlabel('Quantidade')
plt.ylabel('Frequência')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Valor'], bins=10, kde=True)
plt.title('Histograma - Valor')
plt.xlabel('Valor')
plt.ylabel('Frequência')
plt.show()
```
